```python
#First off, the library used will be pandas, therefore, it must be imported
import pandas as pd
```

```python
#Next, the file must be loaded
file_location = '~/assessments/hpdm206Z/assessment2/human_genetic_annotations.txt'

#The file does not contain column names, therefore these should be defined
col_names= ["variant_id", "variant_unit", "annotation"] 

#Data are separated by '\t' (tab), therefore, this should be specified in the code
genetic_data = pd.read_csv(file_location, sep='\t', names=col_names)
```

```python
#Display information about the uploaded dataset, including how many empty / null counts in each column
genetic_data.info(show_counts=True)
```

```python
#Display the first 20 rows in human-readible form
genetic_data.head(20)
```

```python
##Display the first 20 rows in human-readible form
genetic_data.tail(20)
```

```python
#Data cleaning: Firstly, null values must be removed using .dropna()
genetic_data = genetic_data.dropna()
```

```python
#Now, to double check if all missing values were removed:
genetic_data.info(show_counts=True)
```

```python
#Display the unique values in each column, as well as the number of unique values

import numpy as np
print ("variant_id list of unique values: ", genetic_data.variant_id.pipe(np.unique))

#This column must take the form of a string with four colon-separated values, 
#where the first two values are positive integers (refer to the chromosome number, and the position 
#on that chromosome), and the latter two are strings that exclusively contain the 
#characters 'A', 'C', 'G', and 'T' (the four nucleotide bases in humans). 
```

```python
print ("variant_id number of unique values: ", genetic_data.variant_id.nunique())
```

```python
print ("variant_unit list of unique values: ", genetic_data.variant_unit.pipe(np.unique))
```

```python
print ("variant_unit number of unique values: ", genetic_data.variant_unit.nunique())
```

```python
print ("annotation list of unique values: ", genetic_data.annotation.pipe(np.unique))
```
